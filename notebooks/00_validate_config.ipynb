```python
import json
from pathlib import Path

# Identify configuration directory
config_dir = Path.cwd().parent / "config"

# Required configuration parameters
required_keys = {
    "environment",
    "source_system",
    "source_database",
    "source_catalog",
    "source_schema",
    "catalog",
    "bronze_schema",
    "silver_schema",
    "gold_schema",
    "audit_schema",
    "quarantine_schema",
    "warehouse",
    "tables"
}

# Validate each environment
for env in ["dev", "sit", "uat"]:

    config_path = config_dir / f"{env}.json"

    with open(config_path, "r") as file:
        config = json.load(file)

    # Validate required parameters
    missing_keys = required_keys - config.keys()

    if missing_keys:
        raise ValueError(
            f"{env}: Missing parameters {missing_keys}"
        )

    # Validate environment name
    if config["environment"] != env.upper():
        raise ValueError(f"{env}: Invalid environment")

    # Validate target catalog
    if config["catalog"] != f"sales_{env}":
        raise ValueError(f"{env}: Invalid target catalog")

    print(f"{env.upper()} Configuration: PASSED")
    print(f"Source: {config['source_catalog']}.{config['source_schema']}")
    print(f"Target: {config['catalog']}")
    print(f"Number of tables: {len(config['tables'])}")
    print("-" * 50)

print("All environment configurations validated successfully.")
```
